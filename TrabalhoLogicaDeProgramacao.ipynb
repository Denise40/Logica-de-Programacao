```python
print('Bem vindo a Loja da Denise Garcia')
valorPedido = float(input('Entre com o valor do pedido: '))
quantidadeParcelas = int(input('Entre com a quantidade de Parcelas: '))

#implementação dos juros
if quantidadeParcelas < 4:
  juros = 0
elif 4 <=  quantidadeParcelas < 6:
  juros = 0.04
elif 6 <= quantidadeParcelas < 9:
  juros = 0.08
elif 9 <= quantidadeParcelas < 13:
  juros = 0.16
else:
  juros = 0.32

#calculo do valor da parcela
valorParcela = valorPedido * (1 + juros) / quantidadeParcelas

#calculo do valor total
valorTotalParcelado = valorParcela * quantidadeParcelas

#mostrando os resultados
print(f'O valor das parcelas é de: R$ {valorParcela:.2f}')
print(f'O valor total parcelado é de: R$ {valorTotalParcelado:.2f}')
```

```python
#realizando a impressão do menu
print('-' * 8,'Bem vindo a Loja de Marmitas de Denise Garcia', '-' * 7)
print('-' * 26,'CARDÁPIO','-' * 26)
print('-' * 3, '|', 'Tamanho', '|', 'Bife Acebolado(BA)', '|', 'Filé de Frango(FF)', '|', '-' * 4)
print('-' * 3, '|', '   P   ', '|', '     R$16,00      ', '|', '     R$15,00      ', '|', '-' * 4)
print('-' * 3, '|', '   M   ', '|', '     R$18,00      ', '|', '     R$17,00      ', '|', '-' * 4)
print('-' * 3, '|', '   G   ', '|', '     R$22,00      ', '|', '     R$21,00      ', '|', '-' * 4)
print('-' * 62)


#acumulador de pedidos
totalPedido = 0

while True:
  #entrada do sabor
  sabor = input('Entre com o sabor desejado (BA/FF): ').upper()
  if sabor != 'BA' and sabor != 'FF':
    print('Sabor inválido. Tente novamente!\n')
    continue   #volta ao inícion do laço

  #entrada do tamamnho
  tamanho = input('Entre com o tamanho desejado (P/M/G): ').upper()
  if tamanho != 'P' and tamanho != 'M' and tamanho != 'G':
    print('Tamanho inválido. Tente novamente!\n')
    continue  #volta ao inícion do laço

  #calcular o preço de cada item escolhido
  if sabor == 'BA':
    if tamanho == 'P':
      preco = 16
    elif tamanho == 'M':
      preco = 18
    elif tamanho == 'G':
      preco = 22

  elif sabor == 'FF':
    if tamanho == 'P':
      preco = 15
    elif tamanho == 'M':
      preco = 17
    elif tamanho == 'G':
      preco = 21

  #soma dos preços ao total do pedido
  totalPedido += preco

  print(f'Você pediu um {sabor} no tamanho {tamanho} : R${preco:.2F}\n')

#perguntar se deseja continuar
  continuar = input('Deseja mais alguma coisa? (S/N): ').upper()
  if continuar == 'N':
    print(f'\nO valor total a ser pago é de: R${totalPedido:.2f}') #mostra o valor total do pedido
    break #encerra-se o pedido
  elif continuar != 'S': #caso haja resposta que não seja S ou N
    print('Opção inválida. Tente novamente!\n')
    continuar = input('Deseja mais alguma coisa? (S/N): ').upper() #retorna a pargunta
```

```python
#mensagem de boas vindas
print('Bem vindo a Fábrica de Camisetas da Denise Garcia')

#função para escolher o modelo
def escolha_modelo ():
  while True:
    print('\nEntre com o modelo desejado: ')
    print('MCS - Manga Curta Simples')
    print('MLS - Manga Longa Simples')
    print('MCE - Manga Curta Com Estampa')
    print('MLE - Manga Longa Com Estampa')
    modelo = input('>>').upper()

    if modelo == 'MCS':
      return 1.80
    elif modelo == 'MLS':
      return 2.10
    elif modelo == 'MCE':
      return 2.90
    elif modelo == 'MLE':
      return 3.20
    else:
      print('Escolha inválida, entre com o modelo novamente!\n')
      continue


#função para escolher a quantidade de camisas e calcular o desconto
def num_camisetas():
  while True:
      qtd = input('Entre com o número de camisetas: ')
      try:
        qtd = int(qtd)
      except ValueError:
        print('Quantidade inválida. Tente novamente!\n')
        continue
      if qtd > 20000:
        print('Não aceitamos tantas camisetas de uma vez.')
        print('Por favor, entre com o número de camisetas novamente.\n')
        continue
      elif qtd >= 2000:
        return int(qtd * 0.88)  #12% de desconto
      elif qtd >= 200:
        return int(qtd * 0.93)  #7% de desconto
      elif qtd >= 20:
        return int(qtd * 0.95)  #5% de desconto
      elif qtd >0:
        return qtd #sem desconto
      else:
        print('Quantidade inválida. Tente novamente!\n')
        continue

#função para cálculo do frete
def frete():
  while True:
    print('\nEscolha o tipo de frete: ')
    print('0 - Retirar pedido na fábrica - R$ 0.00')
    print('1 - Frete por transportadora - R$ 100.00')
    print('2 - Frete por Sedex - R$ 200.00')

    tipo_frete = input('>>')
    try:
      tipo_frete = int(tipo_frete)
    except ValueError:
      print('Tipo de frete inválido. Escolha entre 0/1/2!\n')
      continue
    if tipo_frete == 0:
      return 0
    elif tipo_frete == 1:
      return 100
    elif tipo_frete == 2:
      return 200
    else:
      print('Opção inválida. Escolha entre 0/1/2!\n')
      continue

#colocando as funções em uma variável
modelo_valor = escolha_modelo()
quantidade_final = num_camisetas()
frete_valor = frete()

#calculo do valor total
total = (modelo_valor * quantidade_final) + frete_valor

#printando a mensagem do total a pagar
print(f'\nTotal: R$ {total:.2f} (Modelo: {modelo_valor:.2f} * Quantidade(com desconto): {quantidade_final} + frete: {frete_valor:.2f})')




```

```python
#lista vazia para armazenar os dados dos funcionários
lista_funcionarios = []

#função para cadastrar funcionário
def cadastrar_funcionario(id_atual):
  while True:
    print('-' * 50)
    print('-'*11, 'MENU CADASTRAR FUNCIONÁRIO', '-'*11)
    print(f'ID do Funcionário: {id_atual}')
    nome = input('Por favor entre com o nome do funcionário: ')
    setor = input('Por favor entre com o setor do funcionário: ').upper()
    salario = float(input('Por favor entre com o salário do funcionário: '))

    #dicionário com chave:valor do funcionário cadastrado
    funcionario = {
        'id': id_atual,
        'nome': nome,
        'setor': setor,
        'salario': salario
    }

    #Adicionando o dicionário na lista vazia
    lista_funcionarios.append(funcionario.copy())

    return id_atual + 1  # retorna o próximo ID

#Função de consultar funcionários cadastrados
def consultar_funcionarios():
  while True:
    print('-' * 50)
    print('-'*11, 'MENU CONSULTAR FUNCIONÁRIO', '-'*11)
    print('Escolha a opção Desejada:')
    print('1 - Consultar todos os Funcionários')
    print('2 - Consultar Funcionário por ID')
    print('3 - Consultar Funcionário por setor')
    print('4 - Retornar ao Menu')
    opcao = input(">>")

    if opcao == '1': #consultar todos os funcionários
      for i in lista_funcionarios:
        print('-' * 30)
        print(f"ID: {i['id']}")
        print(f"Nome: {i['nome']}")
        print(f"Setor: {i['setor']}")
        print(f"Salário: R$ {i['salario']:.2f}")

    elif opcao == '2': #consultar por ID
      buscar_id = int(input('Digite o ID do funcionário: '))
      encontrado = False
      for i in lista_funcionarios:
        if i['id'] == buscar_id:
          print('-' * 30)
          print(f"ID: {i['id']}")
          print(f"Nome: {i['nome']}")
          print(f"Setor: {i['setor']}")
          print(f"Salário: R$ {i['salario']:.2f}")
          encontrado = True
          break
      if not encontrado:
        print('Funcionário não encontrado')

    elif opcao == '3': #consultar por setor
      buscar_setor = input('Digite o setor do funcionário: ').upper()
      encontrado = False
      for i in lista_funcionarios:
        if i['setor'] == buscar_setor:
          print('-' * 10)
          print(f"ID: {i['id']}")
          print(f"Nome: {i['nome']}")
          print(f"Setor: {i['setor']}")
          print(f"Salário: R$ {i['salario']:.2f}")
          encontrado = True
      if not encontrado:
        print('Nenhum funcionário cadastrado nesse setor\n')

    elif opcao == '4': #retornar ao menu
      return
    else:
      print('Opção inválida. Tente novamente!\n') #no caso de não escolher nenhuma das opções corretamente
    continue

#Função de remover funcionário cadastrado
def remover_funcionario():
  while True:
    print('-' * 50)
    print('-'*11, 'MENU REMOVER FUNCIONÁRIO', '-'*11)
    id_remover = int(input("Digite o ID do funcionário a ser removido: "))
    for i in lista_funcionarios:
      if i['id'] == id_remover:
        lista_funcionarios.remove(i)
        print('Funcionário removido com sucesso!\n')
        encontrado = True
        break
    if not encontrado:
        print('ID inválido\n')
    break

#variável global com meu RU
id_global = 5340365

#menu principal no código principal
while True:
  #Mensagem de boas vindas
  print('\nBem-vindo a Empresa de Denise Garcia')
  print('-' * 50)
  print('-'*17, 'MENU PRINCIPAL', '-'*17)
  print('Escolha a opção Desejada:')
  print('1 - Cadastrar Funcionário')
  print('2 - Consultar funcionários')
  print('3 - Remover Funcionário')
  print('4 - Sair')
  opcao = input('>>')

  if opcao == '1':
    id_global = cadastrar_funcionario(id_global)
  elif opcao == '2':
    consultar_funcionarios()
  elif opcao == '3':
    remover_funcionario()
  elif opcao == '4':
    print('Saindo do programa...')
    break
  else:
    print('Opção inválida. Tente novamente!\n')
  continue



```
